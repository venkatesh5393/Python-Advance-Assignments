```python
import numpy as np
# Step 1: Create 1D Array
temperatures_w1 = np.array(
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9]
)

print("Week 1 Temperatures:", temperatures_w1)
```

```python
# Step 2: Properties
print("Shape:", temperatures_w1.shape)
print("Data Type:", temperatures_w1.dtype)
print("Number of Elements:", temperatures_w1.size)
```

```python
# Step 3: Celsius to Fahrenheit
fahrenheit = (temperatures_w1 * 9/5) + 32
print("Fahrenheit:", fahrenheit)
```

```python
# Maximum, Minimum, Mean
print("Maximum:", np.max(temperatures_w1))
print("Minimum:", np.min(temperatures_w1))
print("Mean:", np.mean(temperatures_w1))
```

```python
# Step 4: Slicing
print("First 3 Days:", temperatures_w1[:3])
print("Weekend:", temperatures_w1[-2:])
print("Middle 3 Days:", temperatures_w1[2:5])
```

```python
# Step 5: Create 2D Array
temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]
])

print("2D Array:")
print(temperatures)
```

```python
# Step 6: 2D Array Properties
print("Shape:", temperatures.shape)
print("Data Type:", temperatures.dtype)
print("Total Elements:", temperatures.size)
# Extract each week
print("Week 1:", temperatures[0])
print("Week 2:", temperatures[1])
# Extract weekends
print("Week 1 Weekend:", temperatures[0, -2:])
print("Week 2 Weekend:", temperatures[1, -2:])
```

```python
 import pandas as pd
 # Create Series
marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)

print("Marks:")
print(marks)
```

```python

#1st Rank using integer position
print("1st Rank:", marks.iloc[0])

#Top 3 using loc
print("Top 3 Ranks:")
print(marks.loc[['Rank1', 'Rank2', 'Rank3']])

#3rd Rank using iloc
print("3rd Rank:", marks.iloc[2])

#Marks greater than 90
print("Marks > 90:")
print(marks[marks > 90])
```

```python
#Modify Rank1
marks['Rank1'] = 100

#Remove Rank5
marks = marks.drop('Rank5')

#Calculate CGPA
cgpa = marks / 10

print("Updated Marks:")
print(marks)

print("CGPA:")
print(cgpa)
```

```python
import pandas as pd

#Create DataFrame
transactions = pd.DataFrame({
    'TransactionID': [101,102,103,104,105,106,107,108,109,110],

    'ProductCategory': [
        'Electronics','Clothing','Electronics','Furniture',
        'Clothing','Electronics','Furniture','Clothing',
        'Furniture','Electronics'
    ],

    'Region': [
        'North','South','North','East','West',
        'North','East','West','South','North'
    ],

    'Amount': [200,150,300,450,200,250,300,180,350,400]
})

#Display DataFrame
print(transactions)

```

```python

# Head and Tail
print(transactions.head())
print(transactions.tail())

# Shape, Columns, Data Types
print(transactions.shape)
print(transactions.columns)
print(transactions.dtypes)

# Select columns
print(transactions[['ProductCategory', 'Amount']])

# Last 3 columns
print(transactions.iloc[:, -3:])

# Filter
result = transactions[
    (transactions['Region'] == 'North') &
    (transactions['Amount'] > 200)
]
print(result)

# Value Counts
print(transactions['ProductCategory'].value_counts())

# Unique Regions
print(transactions['Region'].unique())

# GroupBy Mean
print(transactions.groupby('Region')['Amount'].mean())
```

```python
# Modify Transaction 102
transactions.loc[
    transactions['TransactionID'] == 102,
    'Amount'
] = 165

# Add Discount
transactions['Discount'] = transactions['Amount'] * 0.10

# Remove Transaction 109
transactions = transactions[
    transactions['TransactionID'] != 109
]

# Delete Discount
transactions = transactions.drop(columns=['Discount'])

print(transactions)
```
